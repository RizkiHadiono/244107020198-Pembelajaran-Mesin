1. Contoh tindakan yang melanggar etika dan hukum dalam penggunaan kecerdasan buatan

Salah satu contohnya adalah membuat dan menyebarkan deepfake seseorang tanpa izin. Deepfake merupakan konten seperti foto, video, atau suara yang dibuat atau dimanipulasi menggunakan AI sehingga terlihat seperti orang tersebut benar-benar melakukan atau mengatakan sesuatu.

Tindakan tersebut dapat melanggar etika karena dapat merugikan privasi, reputasi, dan hak seseorang. Jika digunakan untuk menipu, menyebarkan informasi palsu, atau merugikan orang lain, tindakan tersebut juga dapat menimbulkan masalah hukum sesuai dengan peraturan yang berlaku.

UNESCO menyatakan bahwa penggunaan AI perlu memperhatikan hak asasi manusia, privasi, perlindungan data, transparansi, serta tanggung jawab manusia.

Contoh: seseorang menggunakan AI untuk membuat video palsu yang menampilkan orang lain seolah-olah melakukan sesuatu yang sebenarnya tidak pernah dilakukan, kemudian menyebarkannya di media sosial tanpa izin. Menurut saya, penggunaan AI seperti ini tidak etis karena dapat merugikan orang lain.

Referensi:


*   UNESCO. Recommendation on the Ethics of Artificial Intelligence.

2. Dampak energi dan lingkungan dari pemanfaatan kecerdasan buatan dan cara mengatasinya

AI membutuhkan komputer dengan kemampuan tinggi untuk melatih dan menjalankan model. Komputer tersebut biasanya berada di pusat data (data center) yang membutuhkan listrik dalam jumlah besar. Selain untuk menjalankan server, energi juga digunakan untuk sistem pendingin dan peralatan pendukung lainnya.

Menurut International Energy Agency (IEA), pusat data menggunakan sekitar 415 TWh listrik pada tahun 2024, atau sekitar 1,5% konsumsi listrik dunia. Dalam skenario dasar IEA, konsumsi listrik pusat data diperkirakan meningkat menjadi sekitar 945 TWh pada tahun 2030. Pertumbuhan AI menjadi salah satu faktor penting yang mendorong peningkatan tersebut.

Penggunaan listrik tersebut juga dapat memberikan dampak terhadap lingkungan karena sebagian listrik masih dihasilkan dari bahan bakar fosil. IEA memperkirakan pusat data menghasilkan sekitar 180 juta ton emisi CO₂ tidak langsung pada tahun 2024 dari penggunaan listrik.

Menurut pendapat saya, dampak tersebut dapat dikurangi dengan beberapa cara, yaitu:

Menggunakan model AI yang lebih efisien sehingga membutuhkan lebih sedikit komputasi.
Menggunakan perangkat keras yang hemat energi.
Menggunakan energi dari sumber terbarukan seperti tenaga surya, angin, dan air.
Melakukan optimasi model agar proses pelatihan tidak membutuhkan komputasi yang berlebihan.
Menggunakan AI sesuai kebutuhan dan menghindari penggunaan sumber daya komputasi secara tidak perlu.

Dengan cara tersebut, perkembangan teknologi AI tetap dapat dimanfaatkan tetapi penggunaan energi dan dampaknya terhadap lingkungan dapat dikurangi. UNESCO juga memasukkan sustainability serta perlindungan lingkungan dan ekosistem sebagai bagian dari prinsip etika AI.

Referensi:

*   International Energy Agency (IEA). AI and Climate Change.
